```python
# Install first (run in terminal once)
# pip install spacy
# python -m spacy download en_core_web_sm

import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = """
Sundar Pichai is the CEO of Google.
He visited Chennai on 5th July 2023.
"""

# Process text
doc = nlp(text)

print("="*50)
print("1️⃣ TOKENS")
print("="*50)
print([token.text for token in doc])


print("\n" + "="*50)
print("2️⃣ REMOVE STOPWORDS & PUNCTUATION")
print("="*50)
filtered_tokens = [
    token.text for token in doc 
    if not token.is_stop and not token.is_punct
]
print(filtered_tokens)


print("\n" + "="*50)
print("3️⃣ LEMMATIZATION")
print("="*50)
for token in doc:
    if not token.is_punct:
        print(f"{token.text}  -->  {token.lemma_}")


print("\n" + "="*50)
print("4️⃣ PART-OF-SPEECH (POS) TAGGING")
print("="*50)
for token in doc:
    if not token.is_punct:
        print(f"{token.text}  -->  {token.pos_}")


print("\n" + "="*50)
print("5️⃣ NAMED ENTITY RECOGNITION (NER)")
print("="*50)
for ent in doc.ents:
    print(f"{ent.text}  -->  {ent.label_}")
```
