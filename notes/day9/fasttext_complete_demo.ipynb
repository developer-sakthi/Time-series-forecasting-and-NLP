## Step 1: Create Multi-Class Training Data


```python
with open('train_mc.txt', 'w') as f:
    f.write('__label__positive I love this phone\n')
    f.write('__label__negative This product is terrible\n')
    f.write('__label__positive Amazing performance and battery\n')
    f.write('__label__negative Worst experience ever\n')

with open('test_mc.txt', 'w') as f:
    f.write('__label__positive This phone is amazing\n')
    f.write('__label__negative I hate this product\n')
```

## Step 2: Train Multi-Class Model


```python
import fasttext

model_mc = fasttext.train_supervised(
    input='train_mc.txt',
    lr=0.1,
    epoch=15,
    dim=100,
    wordNgrams=2,
    minCount=1
)

print('Multi-class Test Results:')
print(model_mc.test('test_mc.txt'))
N, precision, recall = model_mc.test("test_mc.txt")

print('Prediction:')
print(model_mc.predict('This product is amazing'))
```

## Step 3: Create Multi-Label Training Data


```python
with open('train_ml.txt', 'w') as f:
    f.write('__label__comedy __label__romance This movie is funny and romantic\n')
    f.write('__label__action This movie has fight scenes\n')
    f.write('__label__comedy This film is hilarious\n')
    f.write('__label__romance Beautiful love story\n')

with open('test_ml.txt', 'w') as f:
    f.write('__label__comedy __label__romance Funny romantic movie\n')
    f.write('__label__action Amazing fight sequence\n')
```

## Step 4: Train Multi-Label Model


```python
model_ml = fasttext.train_supervised(
    input='train_ml.txt',
    lr=0.1,
    epoch=20,
    dim=100,
    wordNgrams=2,
    loss='ova'
)

print('Multi-label Test Results:')
print(model_ml.test('test_ml.txt'))

N, precision, recall = model_ml.test("test.txt")

print('Multi-label Prediction:')
print(model_ml.predict('This movie is funny and romantic', k=2))
```
