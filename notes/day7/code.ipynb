```python
import spacy
from spacy.pipeline import EntityRuler

# Load English model
nlp = spacy.load("en_core_web_sm")

# Create EntityRuler and add to pipeline
ruler = EntityRuler(nlp)

# Define patterns
patterns = [
    {
        "label": "INVOICE_ID",
        "pattern": [
            {"LOWER": "invoice"},
            {"IS_DIGIT": True}
        ]
    },
    {
        "label": "PHONE_NUMBER",
        "pattern": [
            {"SHAPE": "dddddddddd"}
        ]
    },
    {
        "label": "SKILL",
        "pattern": [
            {"LOWER": "python"}
        ]
    },
    {
        "label": "PRODUCT",
        "pattern": [
            {"POS": "ADJ"},
            {"POS": "NOUN"}
        ]
    }
]

# Add patterns to ruler
ruler.add_patterns(patterns)
nlp.add_pipe(ruler)

# Test text
text = """
Please check invoice 45678.
Call me at 9876543210.
I know Python.
This is a smart phone.
"""

doc = nlp(text)

# Print detected entities
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
