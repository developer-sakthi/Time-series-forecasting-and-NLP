```python
# ==============================
# 1. IMPORT LIBRARIES
# ==============================
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.multiclass import OneVsRestClassifier

from gensim.models import Word2Vec


# ==============================
# 2. SAMPLE DATA
# ==============================

# Binary classification dataset (Spam / Not Spam)
binary_data = pd.DataFrame({
    "text": [
        "Win money now",
        "Limited time offer",
        "Meeting at 10am",
        "Project deadline tomorrow",
        "Claim your prize",
        "Let's discuss the report"
    ],
    "label": [1, 1, 0, 0, 1, 0]  # 1 = Spam, 0 = Not Spam
})

# Multi-class dataset (News category)
multi_class_data = pd.DataFrame({
    "text": [
        "Team won the match",
        "New government policy",
        "AI is transforming tech",
        "Elections coming soon",
        "New smartphone released"
    ],
    "label": ["Sports", "Politics", "Tech", "Politics", "Tech"]
})

# Multi-label dataset
multi_label_data = pd.DataFrame({
    "text": [
        "Romantic emotional movie",
        "Action thriller film",
        "Comedy romantic movie"
    ],
    "labels": [
        ["Romance", "Drama"],
        ["Action", "Thriller"],
        ["Comedy", "Romance"]
    ]
})


# ==============================
# 3. TF-IDF VECTORIZATION
# ==============================

vectorizer = TfidfVectorizer()
X_binary = vectorizer.fit_transform(binary_data["text"])
y_binary = binary_data["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X_binary, y_binary, test_size=0.3, random_state=42
)

# Train binary classifier
binary_model = LogisticRegression()
binary_model.fit(X_train, y_train)

y_pred = binary_model.predict(X_test)

print("\n===== Binary Classification =====")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))


# ==============================
# 4. MULTI-CLASS CLASSIFICATION
# ==============================

X_multi = vectorizer.fit_transform(multi_class_data["text"])
y_multi = multi_class_data["label"]

X_train_mc, X_test_mc, y_train_mc, y_test_mc = train_test_split(
    X_multi, y_multi, test_size=0.3, random_state=42
)

multi_class_model = LogisticRegression()
multi_class_model.fit(X_train_mc, y_train_mc)

y_pred_mc = multi_class_model.predict(X_test_mc)

print("\n===== Multi-Class Classification =====")
print("Confusion Matrix:")
print(confusion_matrix(y_test_mc, y_pred_mc))
print("\nClassification Report:")
print(classification_report(y_test_mc, y_pred_mc))


# ==============================
# 5. MULTI-LABEL CLASSIFICATION
# ==============================

# Convert labels to binary matrix
mlb = MultiLabelBinarizer()
y_multi_label = mlb.fit_transform(multi_label_data["labels"])

X_multi_label = vectorizer.fit_transform(multi_label_data["text"])

X_train_ml, X_test_ml, y_train_ml, y_test_ml = train_test_split(
    X_multi_label, y_multi_label, test_size=0.3, random_state=42
)

multi_label_model = OneVsRestClassifier(LogisticRegression())
multi_label_model.fit(X_train_ml, y_train_ml)

y_pred_ml = multi_label_model.predict(X_test_ml)

print("\n===== Multi-Label Classification =====")
print("Confusion Matrix:")
print(confusion_matrix(y_test_ml.argmax(axis=1), y_pred_ml.argmax(axis=1)))
print("\nClassification Report:")
print(classification_report(y_test_ml, y_pred_ml))


# ==============================
# 6. WORD2VEC EMBEDDINGS
# ==============================

# Tokenize text
sentences = [text.split() for text in binary_data["text"]]

# Train Word2Vec model
w2v_model = Word2Vec(
    sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1
)

# Get embedding for a word
print("\n===== Word2Vec Example =====")
print("Vector for 'Win':")
print(w2v_model.wv["Win"]) 
```
